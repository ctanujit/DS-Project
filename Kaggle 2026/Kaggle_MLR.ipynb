```python
import pandas as pd
import numpy as np
```

```python
import pandas as pd

# Load the dataset
data = pd.read_csv("/content/Features_Train.csv")

# Clean and convert object columns to float
data['X2'] = data['X2'].astype(str).str.replace(',', '', regex=False).astype(float)
data['X9'] = data['X9'].astype(str).str.replace(',', '', regex=False).astype(float)
data['X14'] = data['X14'].astype(str).str.replace(',', '', regex=False).astype(float)

# Save the updated file
data.to_csv('Features_Train_Updated.csv', index=False)
```

```python
df_train_X = pd.read_csv('/content/Features_Train_Updated.csv')
df_train_Y = pd.read_csv('/content/Output_Train.csv')
X_train = np.array(df_train_X.drop(['Id.'], axis=1))
Y_train = np.array(df_train_Y.drop(['Id.'], axis=1))
df_test_X = pd.read_csv('/content/Features_Test.csv')
X_test = np.array(df_test_X.drop(['Id.'], axis=1))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(X_train, Y_train)
Y_pred = lr.predict(X_test)
```

```python
final_output = np.array(Y_pred)
df = pd.DataFrame(final_output, columns=['Happiness'])
df.index += 1
df.index.name = 'Id.'
df.to_csv('Tanujit_Output_Test.csv')
```
